```python
df=df[['id', 'name', 'host_id', 'host_profile_id', 'host_name',
       'neighbourhood', 'latitude', 'longitude',
       'room_type', 'price', 'minimum_nights', 'number_of_reviews',
       'last_review', 'reviews_per_month', 'calculated_host_listings_count',
       'availability_365', 'number_of_reviews_ltm']]
```

```python
df.head()
```

```python
nan_fill = ['price','minimum_nights','number_of_reviews','last_review', 'reviews_per_month',	
            'calculated_host_listings_count','availability_365','number_of_reviews_ltm'] 
df[nan_fill] = df[nan_fill].fillna(0)
```

```python
# Fix data types

df['minimum_nights'] = df['minimum_nights'].astype('Int64')

df['number_of_reviews'] = pd.to_numeric(df['number_of_reviews'], errors='coerce')

df['number_of_reviews'] = df['number_of_reviews'].astype('Int64')

df['last_review'] = pd.to_datetime(df['last_review'],format='%Y-%m-%d',errors='coerce')

df['number_of_reviews_ltm'] = df['number_of_reviews_ltm'].astype('Int64')
```

```python
df.dtypes
```

```python
df.head()
```

```python

# Remove duplicates

n_before = len(df)
df = df.drop_duplicates(subset=['id'], keep='first')
print(f"Duplicates removed: {n_before - len(df)}")
```

```python
#  Standardise data

df['host_name']    = df['host_name'].str.strip().str.title()
df['neighbourhood'] = df['neighbourhood'].str.strip()
```

```python
#  Fix invalid values

df['price'] = df['price'].abs()
```

```python
#  Handle missing values

df['price'] = df['price'].replace(0,df['price'].mean())

df = df.dropna(subset=['host_id'])
```

```python
df.to_csv('airbnb_listing_cleaned.csv', index=False)
```
